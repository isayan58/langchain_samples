```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
# from langchain_core.chatopenai import ChatOpenAI
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser
import os
from dotenv import load_dotenv


load_dotenv()



# generation = input("Enter your generation (e.g., GenZ, Millennial, GenX): ")
```

### ***Chains with Custom Runnables***

```python
# Task 1 - Prompt Template Creation
user_input = input("Enter a topic to start the convo: ")
prompt_template = ChatPromptTemplate.from_messages([("user", "{topic}"),
("system", "You are a {generation} who is like JARVIS from Iron Man. You are a helpful assistant that provides information in a concise and witty manner.")])
```

```python
# Task 2 - Prompt Finalization.

ready_prompt = prompt_template.invoke({"topic": user_input, "generation": "millennial"})
llm_openai= init_chat_model(model=os.environ.get("LLM_MODEL"), temperature = 0.7)
```

```python
# Task 3 - String Output Parser Creation

str_parse = StrOutputParser()
```

```python
def dictionary_parser(text:str)->dict:

    return {"text": text}
```

```python
prompt_post = ChatPromptTemplate.from_messages([("user", "Create a LinkedIn post about {text}"),
("system", "You are a social media influencer. You need to post on LinkedIn based on the text that is provided to you. Just return the post text that can be posted without any pre or post texts.")])
```

```python
# Chain with custom task

chain = prompt_template | llm_openai | str_parse | dictionary_parser | prompt_post | llm_openai | str_parse

chain.invoke({"topic": user_input, "generation": "millennial"})
```
