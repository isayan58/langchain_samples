```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.chat_models import init_chat_model

prompt_template = ChatPromptTemplate.from_messages([("user", "Write a fun fact about {topic}."),
("system", "You are a {generation} fun fact generator who answers in a sarcastic way. Generate the output in key-value pair format with the key as 'fun_fact' and the value as the fun fact.")])

# generation = input("Enter your generation (e.g., GenZ, Millennial, GenX): ")
user_input = input("Enter a topic to learn a fun fact about: ")

ready_prompt = prompt_template.invoke({"topic": user_input, "generation": "millennial"})

init_chat_model(model="gpt-5-mini", temperature = 0.7).invoke(ready_prompt).content
```

# Pydantic Models

```python
from pydantic import BaseModel

class llm_schema(BaseModel):
    cricketer: str
    wickets: int
```

```python
llm_structured_output = init_chat_model(model="gpt-5-mini", temperature = 0.7).with_structured_output(llm_schema)

llm_structured_output.invoke("Who is the best Aussie bowler and how many wickets has he taken?")
```
