```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain.chat_models import init_chat_model

messages = [
    SystemMessage(content="You are a GenZ assistant, who answers in a fun way."),
    HumanMessage(content="Who's the GOAT of Football?"),
]

init_chat_model(model="gpt-5-mini", temperature = 0.7).invoke(messages).content
```

# **Prompts**

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

user_input = input("Enter a topic to learn a fun fact about: ")

dynamic_prompt = ChatPromptTemplate.from_template("Write a fun fact about {topic}.")

ready_prompt = dynamic_prompt.invoke({"topic": user_input})

init_chat_model(model="gpt-5-mini", temperature = 0.7).invoke(ready_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt_template = ChatPromptTemplate.from_messages([("user", "Write a fun fact about {topic}."),
("system", "You are a {generation} fun fact generator who answers in a sarcastic way.")])

generation = input("Enter your generation (e.g., GenZ, Millennial, GenX): ")
user_input = input("Enter a topic to learn a fun fact about: ")

ready_prompt = prompt_template.invoke({"topic": user_input, "generation": generation})

init_chat_model(model="gpt-5-mini", temperature = 0.7).invoke(ready_prompt).content
```
